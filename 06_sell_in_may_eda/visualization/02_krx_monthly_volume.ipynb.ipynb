```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

import numpy as np
import warnings
warnings.simplefilter('ignore')
```

```python
from matplotlib import font_manager, rc
import platform

your_os = platform.system()
if your_os == 'Linux':
    rc('font', family = 'NanumGothic')
elif your_os == 'Windows':
    ttf = "c:/Windows/Fonts/malgun.ttf"
    font_name = font_manager.FontProperties(fname=ttf).get_name()
    rc('font', family = font_name)
elif your_os == 'Darwin':
    rc('font', family = 'AppleGothic')
rc('axes', unicode_minus = False)
```

```python
df1 = pd.DataFrame({'거래량 (천주)':[9781368,9185594,8540895,12212551,10029663,10168377] , '날짜':[202310,202311,202312,202401,202402,202403]})
df1
```

```python
df1['날짜'] = df1['날짜'].astype(str)
df1 = df1.sort_values(by='날짜')
```

```python
plt.figure(figsize=(10, 6))
plt.plot(df1['날짜'], df1['거래량 (천주)'], marker='o', linestyle='-')
plt.xlabel('날짜')
plt.ylabel('거래량 (천주)')
plt.title('유가증권 거래실적')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
