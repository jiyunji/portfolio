## **대한항공 거래량**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

import numpy as np
import warnings
warnings.simplefilter('ignore')
```

```python
from matplotlib import font_manager, rc
import platform

your_os = platform.system()
if your_os == 'Linux':
    rc('font', family = 'NanumGothic')
elif your_os == 'Windows':
    ttf = "c:/Windows/Fonts/malgun.ttf"
    font_name = font_manager.FontProperties(fname=ttf).get_name()
    rc('font', family = font_name)
elif your_os == 'Darwin':
    rc('font', family = 'AppleGothic')
rc('axes', unicode_minus = False)
```

```python
a=pd.read_csv('003490 과거 데이터 (2).csv', encoding='UTF8')
a
```

```python

df=pd.DataFrame(a[['날짜','거래량']])
df
```

```python
df1=df.sort_values(by='날짜')
df1
```

```python
df1['거래량']=df['거래량'].str.replace('M','').astype(float)
df1
```

```python
plt.figure(figsize=(15, 6))
plt.plot(df1['날짜'], df1['거래량'], marker='o', linestyle='-')
plt.xlabel('날짜')
plt.ylabel('거래량')
plt.title('대한항공거래량')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
